**1. Write all the conventions being followed while declaring a variable.**

Conventions and Rules:
1. Must start with a letter (A–Z or a–z) or underscore (_)
2. Can contain only alphanumeric characters and underscores (A–Z, a–z, 0–9, _)
3. Case-sensitive (age, Age, AGE are different)
4. No spaces or special symbols allowed in names
5. No need to declare data type (dynamic typing)
6. Supports multiple assignments in one line
7. Variables should not start with uppercase unless it's a class name

**5. Using an example, explain how the variables in python are dynamic in nature.**

```python
v = 100
print("Initially, value is:", v, " Type:", type(v))

v = "Now, New"
print("After reassignment, value is:", v, " Type:", type(v))

v = [1, 2, 3, 4]
print("Reassigned again, value is:", v, " Type:", type(v))
```
