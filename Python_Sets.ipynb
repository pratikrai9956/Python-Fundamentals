```python
#1
a = {1, 2, 3, 2, 1}
print(a)
print("Type:", type(a))
```

```python
#2
s1 = {10, 20, 30}

s2 = set()

print("Set 1:", s1)
print("Empty Set:", s2)
```

```python
#3
s = {1, 2, 3}
s.add(4)
s.update([5, 6, 7])

print("After adding elements:", s)
```

```python
#4
s = {"a", "b", "c", "d"}

s.remove("b")     # removes b
print("Updated set:", s)
```

```python
#5
s = {100, 200, 300, 400}
print("Length of set:", len(s))
```
