**Q2. Refer the below variable:**
**x = ‘a’**
**Here, is x a character type or string type variable? Support your answer with an explanation.**

```python
x = 'a'
print("Type of x:", type(x))
print("Length of x:", len(x))
```

**Explanation:**
  - Since Python lacks a char type, it stores characters as string.
  - x holds a single character but is internally a string object.
  - len(x) confirms it's a string of size 1.

**Q4. For the variable: word = ‘PanaJi@12256’**
**Calculate the following:**
**(a) Total number of alphabets in lowercase**
**(b) Total number of alphabets in uppercase**
**(c) Total number of numerical in string**

```python
word = "PanaJi@12256"

lc = sum(c.islower() for c in word)
uc = sum(c.isupper() for c in word)
num = sum(c.isdigit() for c in word)

print("(a) Lowercase letters:", lc)
print("(b) Uppercase letters:", uc)
print("(c) Numeric characters:", num)
```
