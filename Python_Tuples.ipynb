```python
#1
t = (1, 2, 3, 2, 1)
print(t)
print("Type:", type(t))
```

```python
#2
t1 = (10, 20, 30)
print(t1)
```

```python
#3
lst = [1, 2, 3]
lst.append(4)  # allowed

tpl = (1, 2, 3)
# tpl.append(4)  # Not allowed → AttributeError
print(lst, tpl)
```

```python
#4
t = (1, 2, [3, 4])
t[2].append(5)  # allowed because list inside tuple is mutable
print(t)
```

```python
#5
t = ("Python", "SQL", "ML")
print("First element:", t[0])
print("Last element:", t[-1])
```

```python
#6
t = ("John", 25, 90.5)
name, age, marks = t  # unpacking

print(name)
print(age)
print(marks)
```
